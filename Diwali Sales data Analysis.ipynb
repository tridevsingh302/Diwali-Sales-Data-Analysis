# Exploratory Data Analysis (EDA)

# Gender

```python
df.columns
```

```python
# count plot of Gender..
ax = sns.countplot(x = 'Gender', data = df, hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_amt = df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values('Amount',ascending = False)
sales_amt
```

```python
# bar chart of Amount spend by Gender(Male and Female)...
ay = sns.barplot(x = 'Gender' , y = 'Amount' , data = sales_amt, legend=False, hue = 'Gender' )
for bars in ay.containers:
    ay.bar_label(bars)
```

**From the above graphs, we see that most of the buyers are female and even the purchasing power of females is greater than mans**

# Age

```python
# count plot of age groups..
plt.figure(figsize = (12,6))
ax = sns.countplot(data = df,x = 'Age Group', hue = 'Gender')
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
# Total Amount v/s Age group...
sales_age = df.groupby(['Age Group'],as_index = False)['Amount'].sum().sort_values('Amount',ascending = False)
sales_age
```

```python
# bar chart of Age_group spending Amount..
sns.barplot(data = sales_age, x = 'Age Group', y = 'Amount' , palette='Set2', hue =  'Age Group')
```

**From the above graph, Most of the buyers are between age of 26-35**

# State

```python
df.columns
```

```python
# total number of orders from top 10 states..
sales_state = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = 'Orders',ascending = False).head(10)
sales_state
```

```python
# bar chart of total number of orders from diff states..
plt.figure(figsize = (15,6))
sns.barplot(data = sales_state,x = 'State', y = 'Orders',palette='Set2',hue = 'State' )
```

```python
# Total amount spent by top 10 states...
Total_amt = df.groupby(['State'], as_index = False)['Amount'].sum().sort_values(by = 'Amount',ascending = False).head(10)
Total_amt
```

```python
# bar chart of total amount spend by states...
plt.figure(figsize = (15,6))
plt.xticks(rotation = 45, ha = 'right')
sns.barplot(data = Total_amt,x = 'State', y = 'Amount',palette='Set2',hue = 'State' )
```

**From above graph we can see that most of the orders are from UttarPradesh, Maharashtra and Karnataka respectively**

# Occupation

```python
df.columns
```

```python
# count plot of occupation sectorss...
plt.figure(figsize = (12,6))
ax = sns.countplot(data = df,x = 'Occupation',palette = 'Set2', hue = 'Occupation')
plt.yticks(rotation = 45, ha = 'right')
plt.xticks(rotation = 45, ha = 'right')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# bar chart of Amount spend by differ occupation peoples..
plt.figure(figsize = (12,6))
plt.xticks(rotation = 45,ha = 'right')
aw = df.groupby(['Occupation'], as_index = False)['Amount'].sum().sort_values('Amount',ascending = False)

ds = sns.barplot(data = aw , x= 'Occupation' , y = 'Amount', palette = 'Set2' , hue = 'Occupation')
```

**Most of the Buyers are from IT, Healthcare and Aviation sectors**

# Product Category

```python
# counter plot of product category
plt.figure(figsize = (12,6))
plt.xticks(rotation = 45,ha = 'right')
ax = sns.countplot(data = df , x= 'Product_Category',palette = 'Set2', hue = 'Product_Category')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# bar chart of Amount spend on differ Product..
plt.figure(figsize = (12,6))
plt.xticks(rotation = 45,ha = 'right')
az = df.groupby(['Product_Category'], as_index = False)['Amount'].sum().sort_values('Amount',ascending = False)
sns.barplot(data = az , x= 'Product_Category', y = 'Amount',palette = 'Set2', hue = 'Product_Category')
```

**From above graph we see that most of the product is sold by Food, Clothings and Electronic shops**
